```python
import gradio as gr
from datetime import datetime

def get_application_text(reciever_name: str, name: str, address: str, date1: str, date2: str) -> str:
    date1 = datetime.fromtimestamp(date1).strftime("%d.%m.%Y")
    date2 = datetime.fromtimestamp(date2).strftime("%d.%m.%Y")

    return f"""
                    {reciever_name}

                    Звернення

    Я, {name}, проживаю за адресою: {address}, повідомляю про відсутність постачання гарячої води у квартирі
    з {date1} до моменту написання звернення.
    Прошу Вас усунути зазначену проблему, відновити централізоване постачання шарячої води та надати інформацію про причини відключення та терміни відновлення послуги.

    Дата: {date2}
    """

demo = gr.Interface(
    fn=get_application_text,
    title="Генератор звернень",
    description="Додаток для автоматичної генерації тексту звернень про відсутність гарячої води.",
    inputs=[
        gr.Textbox(label="Одержувач", value="Київтеплоенерго"),
        gr.Textbox(label="Введіть Ваше ПІБ", placeholder="Наприклад: Бабич Софія Дмитрівна"),
        gr.Textbox(label="Введіть Вашу адресу", placeholder="Наприклад: м. Київ, вул. Хрещатик, 1"),
        gr.DateTime(label="Оберіть дату відключення опалення", include_time=False),
        gr.DateTime(label="Оберіть дату написання звернення", include_time=False)
    ],
    outputs=[
        gr.Textbox(label="Згенерований текст звернення", lines=12)
    ],
    submit_btn="Згенерувати",
    clear_btn="Очистити",
)

demo.launch(share=True)
```
